```python
import torch
from transformers import AutoModelForSequenceClassification
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

# --------------------------------------------------
# Device
# --------------------------------------------------

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)


# --------------------------------------------------
# Model
# --------------------------------------------------

model = AutoModelForSequenceClassification.from_pretrained(
    "google/bert_uncased_L-2_H-128_A-2",
    num_labels=2
)

model = model.to(device)


# --------------------------------------------------
# Optimizer
# --------------------------------------------------

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=2e-5,
    weight_decay=0.01
)


num_epochs = 3

for epoch in range(num_epochs):

    # ==============================================
    # TRAINING
    # ==============================================

    model.train()

    total_train_loss = 0

    train_progress = tqdm(
        train_loader,
        desc=f"Epoch {epoch + 1}/{num_epochs} [Train]",
        leave=True
    )

    for batch in train_progress:

        input_ids = batch["tokenized"].to(device)
        attention_mask = batch["attention_masks"].to(device)
        labels = batch["label"].to(device)

        optimizer.zero_grad()

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            labels=labels
        )

        loss = outputs.loss

        loss.backward()
        optimizer.step()

        total_train_loss += loss.item()

        # Update progress bar
        train_progress.set_postfix(
            loss=f"{loss.item():.4f}"
        )

    avg_train_loss = total_train_loss / len(train_loader)


    # ==============================================
    # VALIDATION
    # ==============================================

    model.eval()

    total_val_loss = 0

    all_labels = []
    all_predictions = []
    all_probabilities = []

    val_progress = tqdm(
        val_loader,
        desc=f"Epoch {epoch + 1}/{num_epochs} [Val]",
        leave=True
    )

    with torch.no_grad():

        for batch in val_progress:

            input_ids = batch["tokenized"].to(device)
            attention_mask = batch["attention_masks"].to(device)
            labels = batch["label"].to(device)

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                labels=labels
            )

            loss = outputs.loss

            total_val_loss += loss.item()

            logits = outputs.logits
            

            probabilities = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            predictions = torch.argmax(
                logits,
                dim=1
            )

            all_labels.extend(
                labels.cpu().numpy()
            )

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_probabilities.extend(
                probabilities.cpu().numpy()
            )

            val_progress.set_postfix(
                loss=f"{loss.item():.4f}"
            )

    avg_val_loss = total_val_loss / len(val_loader)


    # ==============================================
    # VALIDATION METRICS
    # ==============================================

    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    precision = precision_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    recall = recall_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    f1 = f1_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        all_labels,
        all_probabilities
    )

    pr_auc = average_precision_score(
        all_labels,
        all_probabilities
    )


    # ==============================================
    # RESULTS
    # ==============================================

    print(f"\nEpoch {epoch + 1}/{num_epochs}")
    print(f"Train Loss : {avg_train_loss:.4f}")
    print(f"Val Loss   : {avg_val_loss:.4f}")
    print(f"Accuracy   : {accuracy:.4f}")
    print(f"Precision  : {precision:.4f}")
    print(f"Recall     : {recall:.4f}")
    print(f"F1         : {f1:.4f}")
    print(f"ROC-AUC    : {roc_auc:.4f}")
    print(f"PR-AUC     : {pr_auc:.4f}")
```

```python
import time
import numpy as np
import torch

from tqdm.auto import tqdm
from transformers import (
    AutoModelForSequenceClassification,
    AutoTokenizer
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix
)


# ============================================================
# 1. CONFIG
# ============================================================

MODEL_DIR = "./best_bert_classifier"

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print(f"Device: {device}")

if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")

print("=" * 70)


# ============================================================
# 2. LOAD TRAINED MODEL
# ============================================================

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_DIR
).to(device)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_DIR
)

model.eval()

print(f"Loaded model from: {MODEL_DIR}")


# ============================================================
# 3. STORAGE
# ============================================================

all_labels = []
all_predictions = []
all_probabilities = []

total_test_loss = 0.0
total_samples = 0

batch_latencies = []


# ============================================================
# 4. TESTING + LATENCY
# ============================================================

test_progress = tqdm(
    test_loader,
    desc="Testing",
    leave=True
)

with torch.no_grad():

    for batch in test_progress:

        # ----------------------------------------------------
        # Move data to device
        # ----------------------------------------------------

        input_ids = batch["tokenized"].to(device)
        attention_mask = batch["attention_masks"].to(device)
        labels = batch["label"].to(device)

        # ----------------------------------------------------
        # Start timing
        # ----------------------------------------------------

        if device.type == "cuda":
            torch.cuda.synchronize()

        start_time = time.perf_counter()

        # ----------------------------------------------------
        # Model inference
        # ----------------------------------------------------

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            labels=labels
        )

        # ----------------------------------------------------
        # Stop timing
        # ----------------------------------------------------

        if device.type == "cuda":
            torch.cuda.synchronize()

        end_time = time.perf_counter()

        batch_latency = end_time - start_time

        batch_latencies.append(batch_latency)

        # ----------------------------------------------------
        # Get predictions
        # ----------------------------------------------------

        loss = outputs.loss
        logits = outputs.logits

        probabilities = torch.softmax(
            logits,
            dim=1
        )[:, 1]

        predictions = torch.argmax(
            logits,
            dim=1
        )

        # ----------------------------------------------------
        # Store results
        # ----------------------------------------------------

        batch_size = labels.size(0)

        total_test_loss += loss.item() * batch_size
        total_samples += batch_size

        all_labels.extend(
            labels.cpu().numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_probabilities.extend(
            probabilities.cpu().numpy()
        )

        test_progress.set_postfix(
            latency=f"{batch_latency * 1000:.2f} ms"
        )


# ============================================================
# 5. CLASSIFICATION METRICS
# ============================================================

avg_test_loss = total_test_loss / total_samples

accuracy = accuracy_score(
    all_labels,
    all_predictions
)

precision = precision_score(
    all_labels,
    all_predictions,
    zero_division=0
)

recall = recall_score(
    all_labels,
    all_predictions,
    zero_division=0
)

f1 = f1_score(
    all_labels,
    all_predictions,
    zero_division=0
)

roc_auc = roc_auc_score(
    all_labels,
    all_probabilities
)

pr_auc = average_precision_score(
    all_labels,
    all_probabilities
)


# ============================================================
# 6. LATENCY METRICS
# ============================================================

batch_latencies = np.array(batch_latencies)

# Average latency per batch
avg_batch_latency = np.mean(batch_latencies)

# Median latency per batch
median_batch_latency = np.median(batch_latencies)

# 95th percentile latency
p95_batch_latency = np.percentile(
    batch_latencies,
    95
)

# 99th percentile latency
p99_batch_latency = np.percentile(
    batch_latencies,
    99
)

# Total inference time
total_inference_time = np.sum(
    batch_latencies
)

# Average latency per sample
avg_sample_latency = (
    total_inference_time / total_samples
)

# Throughput
throughput = (
    total_samples / total_inference_time
)


# Convert seconds -> milliseconds

avg_batch_latency_ms = (
    avg_batch_latency * 1000
)

median_batch_latency_ms = (
    median_batch_latency * 1000
)

p95_batch_latency_ms = (
    p95_batch_latency * 1000
)

p99_batch_latency_ms = (
    p99_batch_latency * 1000
)

avg_sample_latency_ms = (
    avg_sample_latency * 1000
)


# ============================================================
# 7. FINAL TEST RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("FINAL TEST RESULTS")
print("=" * 70)

print("\nClassification Metrics")
print("-" * 70)

print(f"Test Loss      : {avg_test_loss:.4f}")
print(f"Accuracy       : {accuracy:.4f}")
print(f"Precision      : {precision:.4f}")
print(f"Recall         : {recall:.4f}")
print(f"F1             : {f1:.4f}")
print(f"ROC-AUC        : {roc_auc:.4f}")
print(f"PR-AUC         : {pr_auc:.4f}")

print("\nLatency Metrics")
print("-" * 70)

print(f"Average Batch Latency    : {avg_batch_latency_ms:.2f} ms")
print(f"Average Sample Latency   : {avg_sample_latency_ms:.2f} ms")
print(f"Median Batch Latency     : {median_batch_latency_ms:.2f} ms")
print(f"P95 Batch Latency        : {p95_batch_latency_ms:.2f} ms")
print(f"P99 Batch Latency        : {p99_batch_latency_ms:.2f} ms")

print(f"Total Inference Time     : {total_inference_time:.4f} sec")
print(f"Throughput               : {throughput:.2f} samples/sec")
print(f"Total Test Samples       : {total_samples}")


# ============================================================
# 8. CLASSIFICATION REPORT
# ============================================================

print("\n")
print("=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        all_labels,
        all_predictions,
        digits=4,
        zero_division=0
    )
)


# ============================================================
# 9. CONFUSION MATRIX
# ============================================================

print("=" * 70)
print("CONFUSION MATRIX")
print("=" * 70)

cm = confusion_matrix(
    all_labels,
    all_predictions
)

print(cm)
```

```python
import numpy as np
from torch.utils.data import Subset, DataLoader


# ============================================================
# CONFIG
# ============================================================

n = 5
batch_size = 100
seed = 42


# ============================================================
# GET COMPLETE PROMPT/DOCUMENT IDs
# ============================================================

train_ids = sampler.get_split('train')

print("Total training prompts:", len(train_ids))


# ============================================================
# SHUFFLE PROMPT IDs
# ============================================================

rng = np.random.default_rng(seed)

train_ids = np.array(train_ids)

rng.shuffle(train_ids)


# ============================================================
# SPLIT PROMPT IDs INTO N SUBSETS
# ============================================================

train_id_subsets = np.array_split(
    train_ids,
    n
)


# ============================================================
# CONVERT PROMPT IDs -> DATASET POSITIONS
#
# This is the important part.
#
# ids_to_position() should return ALL chunks belonging
# to each prompt.
# ============================================================

train_subsets = []

for i, ids in enumerate(train_id_subsets):

    positions = dataset.ids_to_position(
        ids.tolist()
    )

    subset = Subset(
        dataset,
        positions
    )

    train_subsets.append(subset)

    print(
        f"BERT {i + 1}: "
        f"{len(ids)} prompts -> "
        f"{len(subset)} chunks"
    )


# ============================================================
# CREATE DATALOADERS
# ============================================================

ensemble_train_loaders = []

for subset in train_subsets:

    loader = DataLoader(
        subset,
        shuffle=True,
        batch_size=batch_size
    )

    ensemble_train_loaders.append(loader)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("ENSEMBLE TRAINING SUBSETS")
print("=" * 60)

print(
    f"Original training prompts : {len(train_ids)}"
)

print(
    f"Number of BERTs            : {n}"
)

for i, subset in enumerate(train_subsets):

    print(
        f"BERT {i + 1:<2}                  : "
        f"{len(subset)} chunks"
    )
```

```python
import torch
from transformers import AutoModelForSequenceClassification
from tqdm.auto import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

num_models = 5
num_epochs = 5

ensemble_models = []

for model_idx, train_loader in enumerate(ensemble_train_loaders):

    print("\n" + "=" * 70)
    print(f"TRAINING BERT {model_idx + 1}/{num_models}")
    print("=" * 70)

    # Create a completely independent model
    model = AutoModelForSequenceClassification.from_pretrained(
        "google/bert_uncased_L-2_H-128_A-2",
        num_labels=2
    ).to(device)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=2e-5,
        weight_decay=0.01
    )

    best_val_f1 = 0.0

    for epoch in range(num_epochs):

        # --------------------------------------------------
        # Training
        # --------------------------------------------------
        model.train()

        train_loss = 0.0

        progress_bar = tqdm(
            train_loader,
            desc=f"BERT {model_idx + 1} | Epoch {epoch + 1}/{num_epochs}"
        )

        for batch in progress_bar:

            input_ids = batch["tokenized"].to(device)
            attention_mask = batch["attention_masks"].to(device)
            labels = batch["label"].to(device)

            optimizer.zero_grad()

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                labels=labels
            )

            loss = outputs.loss

            loss.backward()
            optimizer.step()

            train_loss += loss.item()

            progress_bar.set_postfix(
                loss=f"{loss.item():.4f}"
            )

        avg_train_loss = train_loss / len(train_loader)

        # --------------------------------------------------
        # Validation
        # --------------------------------------------------
        model.eval()

        val_loss = 0.0
        all_labels = []
        all_predictions = []
        all_probabilities = []

        with torch.no_grad():

            for batch in val_loader:

                input_ids = batch["tokenized"].to(device)
                attention_mask = batch["attention_masks"].to(device)
                labels = batch["label"].to(device)

                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    labels=labels
                )

                loss = outputs.loss
                logits = outputs.logits

                probabilities = torch.softmax(
                    logits,
                    dim=1
                )[:, 1]

                predictions = torch.argmax(
                    logits,
                    dim=1
                )

                val_loss += loss.item()

                all_labels.extend(labels.cpu().numpy())
                all_predictions.extend(predictions.cpu().numpy())
                all_probabilities.extend(probabilities.cpu().numpy())

        avg_val_loss = val_loss / len(val_loader)

        val_accuracy = accuracy_score(
            all_labels,
            all_predictions
        )

        val_precision = precision_score(
            all_labels,
            all_predictions,
            zero_division=0
        )

        val_recall = recall_score(
            all_labels,
            all_predictions,
            zero_division=0
        )

        val_f1 = f1_score(
            all_labels,
            all_predictions,
            zero_division=0
        )

        val_roc_auc = roc_auc_score(
            all_labels,
            all_probabilities
        )

        val_pr_auc = average_precision_score(
            all_labels,
            all_probabilities
        )

        print(
            f"Epoch {epoch + 1}/{num_epochs} | "
            f"Train Loss: {avg_train_loss:.4f} | "
            f"Val Loss: {avg_val_loss:.4f} | "
            f"Val F1: {val_f1:.4f}"
        )

        # Keep the best model for this BERT
        if val_f1 > best_val_f1:

            best_val_f1 = val_f1

            # Save the model's state in memory
            best_model_state = {
                k: v.cpu().clone()
                for k, v in model.state_dict().items()
            }

    # Restore best validation checkpoint
    model.load_state_dict(best_model_state)
    model.to(device)

    ensemble_models.append(model)

    print(f"\nBERT {model_idx + 1} best validation F1: {best_val_f1:.4f}")

print("\nAll 3 BERTs have been trained.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

individual_results = []

for model_idx, model in enumerate(ensemble_models):

    model.eval()

    all_labels = []
    all_predictions = []
    all_probabilities = []

    with torch.no_grad():

        for batch in val_loader:

            input_ids = batch["tokenized"].to(device)
            attention_mask = batch["attention_masks"].to(device)
            labels = batch["label"].to(device)

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            logits = outputs.logits

            probabilities = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            predictions = torch.argmax(
                logits,
                dim=1
            )

            all_labels.extend(labels.cpu().numpy())
            all_predictions.extend(predictions.cpu().numpy())
            all_probabilities.extend(probabilities.cpu().numpy())

    results = {
        "accuracy": accuracy_score(
            all_labels,
            all_predictions
        ),
        "precision": precision_score(
            all_labels,
            all_predictions,
            zero_division=0
        ),
        "recall": recall_score(
            all_labels,
            all_predictions,
            zero_division=0
        ),
        "f1": f1_score(
            all_labels,
            all_predictions,
            zero_division=0
        ),
        "roc_auc": roc_auc_score(
            all_labels,
            all_probabilities
        ),
        "pr_auc": average_precision_score(
            all_labels,
            all_probabilities
        )
    }

    individual_results.append(results)

    print(
        f"BERT {model_idx + 1}: "
        f"Accuracy={results['accuracy']:.4f}, "
        f"Precision={results['precision']:.4f}, "
        f"Recall={results['recall']:.4f}, "
        f"F1={results['f1']:.4f}, "
        f"ROC-AUC={results['roc_auc']:.4f}, "
        f"PR-AUC={results['pr_auc']:.4f}"
    )
```

```python
import numpy as np

all_labels = []
model_probabilities = [
    [] for _ in range(len(ensemble_models))
]

# --------------------------------------------------
# Get predictions from every model
# --------------------------------------------------

for model_idx, model in enumerate(ensemble_models):

    model.eval()

    with torch.no_grad():

        for batch in val_loader:

            input_ids = batch["tokenized"].to(device)
            attention_mask = batch["attention_masks"].to(device)

            if model_idx == 0:
                labels = batch["label"]

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            logits = outputs.logits

            probabilities = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            model_probabilities[model_idx].extend(
                probabilities.cpu().numpy()
            )

# Labels only need to be collected once
for batch in val_loader:
    all_labels.extend(
        batch["label"].numpy()
    )

# --------------------------------------------------
# Average probabilities
# --------------------------------------------------

model_probabilities = np.array(model_probabilities)

ensemble_probabilities = model_probabilities.mean(
    axis=0
)

# --------------------------------------------------
# Convert probability → prediction
# --------------------------------------------------

ensemble_predictions = (
    ensemble_probabilities >= 0.5
).astype(int)

# --------------------------------------------------
# Metrics
# --------------------------------------------------

ensemble_results = {
    "accuracy": accuracy_score(
        all_labels,
        ensemble_predictions
    ),
    "precision": precision_score(
        all_labels,
        ensemble_predictions,
        zero_division=0
    ),
    "recall": recall_score(
        all_labels,
        ensemble_predictions,
        zero_division=0
    ),
    "f1": f1_score(
        all_labels,
        ensemble_predictions,
        zero_division=0
    ),
    "roc_auc": roc_auc_score(
        all_labels,
        ensemble_probabilities
    ),
    "pr_auc": average_precision_score(
        all_labels,
        ensemble_probabilities
    )
}

print("\n" + "=" * 70)
print("3-BERT ENSEMBLE VALIDATION RESULTS")
print("=" * 70)

for metric, value in ensemble_results.items():
    print(f"{metric.upper():10s}: {value:.4f}")
```

```python
import time
import numpy as np
import torch

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

# ============================================================
# FINAL TEST EVALUATION + LATENCY — 3-BERT ENSEMBLE
# ============================================================

print("\n" + "=" * 70)
print("FINAL TEST EVALUATION — 3-BERT ENSEMBLE")
print("=" * 70)

# Make sure all models are in evaluation mode
for model in ensemble_models:
    model.eval()

all_labels = []
all_predictions = []
all_probabilities = []

batch_latencies = []

total_inference_time = 0.0
total_samples = 0

# ------------------------------------------------------------
# Test inference
# ------------------------------------------------------------

with torch.no_grad():

    for batch in test_loader:

        input_ids = batch["tokenized"].to(device)
        attention_mask = batch["attention_masks"].to(device)
        labels = batch["label"].to(device)

        # Synchronize CUDA before timing
        if device.type == "cuda":
            torch.cuda.synchronize()

        start_time = time.perf_counter()

        # ----------------------------------------------------
        # Run all 3 BERTs
        # ----------------------------------------------------

        model_probabilities = []

        for model in ensemble_models:

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            logits = outputs.logits

            probabilities = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            model_probabilities.append(probabilities)

        # ----------------------------------------------------
        # Average predictions from all BERTs
        # ----------------------------------------------------

        ensemble_probability = torch.stack(
            model_probabilities,
            dim=0
        ).mean(dim=0)

        predictions = (
            ensemble_probability >= 0.5
        ).long()

        # Synchronize CUDA after timing
        if device.type == "cuda":
            torch.cuda.synchronize()

        end_time = time.perf_counter()

        # ----------------------------------------------------
        # Record latency
        # ----------------------------------------------------

        batch_latency = end_time - start_time

        batch_latencies.append(batch_latency)
        total_inference_time += batch_latency

        # ----------------------------------------------------
        # Store predictions
        # ----------------------------------------------------

        all_labels.extend(
            labels.cpu().numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_probabilities.extend(
            ensemble_probability.cpu().numpy()
        )

        total_samples += labels.size(0)


# ============================================================
# METRICS
# ============================================================

all_labels = np.array(all_labels)
all_predictions = np.array(all_predictions)
all_probabilities = np.array(all_probabilities)

test_accuracy = accuracy_score(
    all_labels,
    all_predictions
)

test_precision = precision_score(
    all_labels,
    all_predictions,
    zero_division=0
)

test_recall = recall_score(
    all_labels,
    all_predictions,
    zero_division=0
)

test_f1 = f1_score(
    all_labels,
    all_predictions,
    zero_division=0
)

test_roc_auc = roc_auc_score(
    all_labels,
    all_probabilities
)

test_pr_auc = average_precision_score(
    all_labels,
    all_probabilities
)


# ============================================================
# LATENCY STATISTICS
# ============================================================

batch_latencies = np.array(batch_latencies)

average_batch_latency = np.mean(
    batch_latencies
)

median_batch_latency = np.median(
    batch_latencies
)

p95_batch_latency = np.percentile(
    batch_latencies,
    95
)

p99_batch_latency = np.percentile(
    batch_latencies,
    99
)

average_sample_latency = (
    total_inference_time / total_samples
)

throughput = (
    total_samples / total_inference_time
)


# ============================================================
# PRINT RESULTS
# ============================================================

print("\n" + "=" * 70)
print("3-BERT ENSEMBLE — TEST RESULTS")
print("=" * 70)

print(f"Accuracy : {test_accuracy:.4f}")
print(f"Precision: {test_precision:.4f}")
print(f"Recall   : {test_recall:.4f}")
print(f"F1       : {test_f1:.4f}")
print(f"ROC-AUC  : {test_roc_auc:.4f}")
print(f"PR-AUC   : {test_pr_auc:.4f}")

print("\n" + "-" * 70)
print("3-BERT ENSEMBLE — INFERENCE LATENCY")
print("-" * 70)

print(f"Number of BERTs          : {len(ensemble_models)}")
print(f"Batch size               : {test_loader.batch_size}")
print(f"Total test samples       : {total_samples}")
print(f"Total inference time     : {total_inference_time:.4f} seconds")

print(
    f"Average batch latency    : "
    f"{average_batch_latency * 1000:.2f} ms"
)

print(
    f"Average sample latency   : "
    f"{average_sample_latency * 1000:.2f} ms"
)

print(
    f"Median batch latency     : "
    f"{median_batch_latency * 1000:.2f} ms"
)

print(
    f"P95 batch latency        : "
    f"{p95_batch_latency * 1000:.2f} ms"
)

print(
    f"P99 batch latency        : "
    f"{p99_batch_latency * 1000:.2f} ms"
)

print(
    f"Throughput               : "
    f"{throughput:.2f} samples/sec"
)

print("=" * 70)
```
